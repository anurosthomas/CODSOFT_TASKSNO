```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(10,6))
df['City'].value_counts().head(10).plot(kind='bar', color='skyblue')
plt.title('Top 10 Cities by Number of Restaurants')
plt.xlabel('City')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

```python
df['Has Online delivery'].value_counts().plot(kind='pie', autopct='%1.1f%%', colors=['lightgreen','salmon'])
plt.title('Online Delivery Availability')
plt.ylabel('')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['Aggregate rating'], bins=20, color='orange', edgecolor='black')
plt.title('Distribution of Aggregate Ratings')
plt.xlabel('Rating')
plt.ylabel('Frequency')
plt.show()
```

```python
avg_by_price = df.groupby('Price range')['Aggregate rating'].mean()
avg_by_price.plot(kind='line', marker='o', color='purple')
plt.title('Average Rating by Price Range')
plt.xlabel('Price Range')
plt.ylabel('Average Rating')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['Votes'], df['Aggregate rating'], alpha=0.5, color='teal')
plt.title('Votes vs Aggregate Rating')
plt.xlabel('Votes')
plt.ylabel('Aggregate Rating')
plt.show()
```
